```python
import pandas as pd
```

```python
data=pd.read_csv(r'C:\Users\yashm\Desktop\dsbda\movie.csv')
```

```python
data.columns
```

```python
features = ['genres', 'keywords', 'cast', 'director']
for feature in features:
    data[feature] = data[feature].fillna('')
    

def combine_features(row):
    return row['genres'] + " " + row['keywords'] + " " + row['cast'] + " " + row['director']

data["combined_features"] = data.apply(combine_features, axis=1)


from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()
feature_vectors = vectorizer.fit_transform(data["combined_features"])


from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(feature_vectors)
```

```python
def recommend(movie_name):
    movie_name = movie_name.lower()
    
    # Find index
    index = data[data['title'].str.lower() == movie_name].index
    
    if len(index) == 0:
        print("Movie not found")
        return
    
    index = index[0]
    
    # Get similarity scores
    scores = list(enumerate(similarity[index]))
    
    # Sort movies
    sorted_movies = sorted(scores, key=lambda x: x[1], reverse=True)
    
    print("\nTop Recommendations:\n")
    for i, movie in enumerate(sorted_movies[1:6]):
        print(i+1, ".", data.iloc[movie[0]]['title'])
```

```python
recommend("cars")
```
