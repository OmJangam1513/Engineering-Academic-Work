```python
n = 0
board = []
col = []
posDiag = []
negDiag = []
countSolutions = 0


# Print the chessboard
def printBoard():
    for i in range(n):
        for j in range(n):
            print(board[i][j], end=" ")
        print()
    print()


# Backtracking function
def backtrack(r):
    global countSolutions

    # All queens have been placed
    if r == n:
        countSolutions += 1
        print(f"Solution {countSolutions}:")
        printBoard()
        return

    # If the first queen is already placed, skip its row
    if r == 0:
        backtrack(r + 1)
        return

    for c in range(n):
        # Check column and diagonals
        if col[c] or posDiag[r + c] or negDiag[r - c + n - 1]:
            continue

        # Place queen
        col[c] = True
        posDiag[r + c] = True
        negDiag[r - c + n - 1] = True
        board[r][c] = '1'

        # Move to next row
        backtrack(r + 1)

        # Backtrack: remove queen
        col[c] = False
        posDiag[r + c] = False
        negDiag[r - c + n - 1] = False
        board[r][c] = '0'


if __name__ == "__main__":
    n = int(input("Enter the value of N: "))

    if n <= 0:
        print("N must be a positive integer.")
        exit()

    # Initialize chessboard and tracking arrays
    board = [['0' for _ in range(n)] for _ in range(n)]
    col = [False] * n
    posDiag = [False] * (2 * n - 1)
    negDiag = [False] * (2 * n - 1)

    # Input the first queen's position
    row = int(input(f"Enter the row for the first queen (1 to {n}): "))
    column = int(input(f"Enter the column for the first queen (1 to {n}): "))

    # Validate the position
    if row < 1 or row > n or column < 1 or column > n:
        print("Invalid queen position!")
        exit()

    # Convert to zero-based indexing
    row -= 1
    column -= 1

    # Place the first queen
    board[row][column] = '1'
    col[column] = True
    posDiag[row + column] = True
    negDiag[row - column + n - 1] = True

    print(
        f"\nSolutions with the first queen fixed at "
        f"({row + 1}, {column + 1}):\n"
    )

    # The original row-by-row approach supports a fixed queen in row 1
    if row == 0:
        backtrack(0)
    else:
        print(
            "Please place the first queen in row 1 "
            "for this version of the program."
        )
        exit()

    if countSolutions == 0:
        print("No solutions found for this position.")
    else:
        print(f"Total solutions: {countSolutions}")
```
