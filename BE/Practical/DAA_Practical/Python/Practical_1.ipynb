```python
import matplotlib.pyplot as plt


def recursive_fibonacci(n):
    if n <= 1:
        return n
    else:
        return recursive_fibonacci(n - 1) + recursive_fibonacci(n - 2)


def non_recursive_fibonacci(n):
    if n <= 0:
        return

    first = 0
    second = 1
    print(first)

    if n == 1:
        return

    print(second)

    while n - 2 > 0:
        third = first + second
        first = second
        second = third
        print(third)
        n -= 1


def visualize_complexity():
    n_values = list(range(1, 21))

    # Time complexity (theoretical growth)
    recursive_time = [2**n for n in n_values]
    iterative_time = n_values

    # Auxiliary space complexity
    recursive_space = n_values
    iterative_space = [1] * len(n_values)

    # Create two graphs
    fig, axes = plt.subplots(1, 2, figsize=(13, 5))

    # Time complexity graph
    axes[0].plot(
        n_values, recursive_time,
        marker="o", label="Recursive O(2^n)"
    )
    axes[0].plot(
        n_values, iterative_time,
        marker="s", label="Iterative O(n)"
    )

    axes[0].set_yscale("log")
    axes[0].set_title("Time Complexity")
    axes[0].set_xlabel("Input Size (n)")
    axes[0].set_ylabel("Theoretical Growth (log scale)")
    axes[0].legend()
    axes[0].grid(True, linestyle="--", alpha=0.5)

    # Space complexity graph
    axes[1].plot(
        n_values, recursive_space,
        marker="o", label="Recursive O(n)"
    )
    axes[1].plot(
        n_values, iterative_space,
        marker="s", label="Iterative O(1)"
    )

    axes[1].set_title("Auxiliary Space Complexity")
    axes[1].set_xlabel("Input Size (n)")
    axes[1].set_ylabel("Space Growth")
    axes[1].legend()
    axes[1].grid(True, linestyle="--", alpha=0.5)

    plt.tight_layout()
    plt.show()


if __name__ == "__main__":
    n = 10

    print("Recursive Method:")
    for i in range(n):
        print(recursive_fibonacci(i))

    print("\nIterative Method:")
    non_recursive_fibonacci(n)

    visualize_complexity()
```
