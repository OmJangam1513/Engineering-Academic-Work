```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sym
```

```python
def objective(x):
    return (x + 3) ** 2

def derivative(x):
    return 2 * (x + 3)

# Display the symbolic derivative
x_sym = sym.symbols('x')
expr = (x_sym + 3) ** 2
print('Derivative:', sym.diff(expr, x_sym))
```

```python
# Initialize parameters
x_old = 2
rate = 0.1
precision = 0.000001

previous_step_size = 1
max_iters = 1000
iters = 0
x_history = [x_old]

# Gradient descent loop
while previous_step_size > precision and iters < max_iters:
    
    x_new = x_old - rate * derivative(x_old)

    previous_step_size = abs(x_new - x_old)
    x_history.append(x_new)
    x_old = x_new
    iters += 1

print('Local minimum occurs at x =', x_old)
print('Minimum function value =', objective(x_old))
print('Iterations =', iters)
```
