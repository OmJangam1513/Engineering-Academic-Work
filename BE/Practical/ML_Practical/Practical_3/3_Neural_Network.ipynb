```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout
```

```python
df = pd.read_csv("E:\BE\Practical\ML_Practical\Practical_3\Churn_Modelling.csv", index_col = 'RowNumber')
```

```python
df.head()
```

```python
X = df.drop(
    ['Exited','CustomerId', 'Surname'],
    axis=1
)

Y = df['Exited']

X = pd.get_dummies(
    X,
    columns=['Geography', 'Gender'],
    drop_first=True,
    dtype=int
)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size = 0.2, random_state = 0)
```

```python
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# %pip install tensorflow if not installed
```

```python
classifier = Sequential()

# Add input layer and first hidden layer
classifier.add(Input(shape=(X_train.shape[1],)))
classifier.add(Dense(6, activation='relu'))
classifier.add(Dropout(rate=0.1))

# Add second hidden layer
classifier.add(Dense(6, activation='relu'))
classifier.add(Dropout(rate=0.1))
```

```python
# Add output layer
classifier.add(Dense(1, activation='sigmoid'))
```

```python
#Let us take a look at our network
classifier.summary()
```

```python
#Optimize the weights
classifier.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
```

```python
#Fitting the Neural Network
history = classifier.fit(X_train, y_train, batch_size = 32, epochs = 200, validation_split = 0.1, verbose = 1)
```

```python
y_pred_prob = classifier.predict(X_test, verbose=1)

#Converting the probability into binary classes i.e (0,1)
y_pred = (y_pred_prob > 0.5).astype(int).ravel()
```

```python
accuracy_score(y_test, y_pred)
```

```python
cm = confusion_matrix(y_test, y_pred)
print("\nConfusion Matrix:")
print(cm)
```

```python
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```
